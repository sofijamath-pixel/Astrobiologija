## Voda i altermativni rastvaraci

### 1. Ključni princip: Zašto je voda toliko važna za život?

Voda je jedan od najvažnijih molekula za život kakav poznajemo na Zemlji. Njena posebna svojstva proizlaze iz njene molekulske strukture, odnosno iz činjenice da je polaran molekul.

Molekul vode sastoji se od jednog atoma kiseonika i dva atoma vodonika. Kiseonik ima veću elektronegativnost od vodonika, što znači da jače privlači zajedničke elektrone. Zbog toga kiseonik dobija delimično negativno naelektrisanje, dok atomi vodonika imaju delimično pozitivno naelektrisanje. Iako je molekul vode kao celina električno neutralan, ova neravnomerna raspodela naelektrisanja omogućava privlačenje između susednih molekula vode.

Ta privlačenja nazivamo vodoničnim vezama. One nisu toliko jake kao kovalentne veze unutar molekula, ali imaju veoma značajan kolektivni efekat i odgovorne su za mnoga neobična svojstva vode.

Jedno od najvažnijih svojstava je njena sposobnost rastvaranja. Pošto je voda polarna, u njoj se dobro rastvaraju joni i mnogi polarni molekuli. Na primer, kada kuhinjsku so dodamo u vodu, molekuli vode okružuju jone natrijuma i hlora i omogućavaju njihovo razdvajanje i raspoređivanje u rastvoru. Sa druge strane, nepolarne supstance, poput ulja, ne rastvaraju se dobro u vodi.

Voda ima i veoma veliki specifični toplotni kapacitet. To znači da može da primi ili oslobodi relativno veliku količinu toplote, a da joj se temperatura ne promeni mnogo. Takođe, ima veliku toplotu isparavanja, pa je za prelazak vode iz tečnog u gasovito stanje potrebno dosta energije. Još jedna zanimljiva osobina je da je led manje gust od tečne vode, zbog čega pluta na površini. To omogućava da se ispod površinskog sloja leda u jezerima održi tečna voda i da organizmi prežive tokom hladnih perioda.

Kao ilustraciju ovog principa izabrala bih Figure 2.13 iz OpenStax udžbenika *Biology 2e*, koja prikazuje polarnost molekula vode i vodonične veze između susednih molekula. Na slici se vidi kako se delimično pozitivni atomi vodonika jednog molekula privlače sa delimično negativnim atomima kiseonika drugog molekula. Ova jednostavna interakcija predstavlja osnovu velikog broja osobina koje vodu čine pogodnom za život.

### 2. Primer funkcije vode u zemaljskom životu

Jedan od najboljih primera značaja vode jeste njen uticaj na biohemijske procese unutar ćelije.

Većina biohemijskih reakcija u organizmima odvija se u vodenoj sredini, odnosno u citoplazmi ćelija. Voda omogućava rastvaranje i transport različitih supstanci, kao što su joni, šećeri i aminokiseline, koje ćelije koriste za svoj metabolizam.

Na primer, enzimi, koji ubrzavaju biohemijske reakcije, funkcionišu u određenim uslovima temperature i pH vrednosti. Voda omogućava održavanje takve sredine i učestvuje u brojnim hemijskim reakcijama. Bez odgovarajućeg rastvarača, molekuli ne bi mogli dovoljno slobodno da se kreću, međusobno stupaju u interakcije i formiraju složene strukture neophodne za život.

Značaj vode vidimo i na nivou čitavih organizama. Kod ljudi, isparavanje znoja omogućava hlađenje tela, dok kod biljaka kohezija i adhezija molekula vode omogućavaju transport vode od korena do listova.

Dakle, voda nije samo sredina u kojoj se život odvija, već aktivno učestvuje u održavanju njegove strukture i funkcionisanja.

### 3. Astrobiološka hipoteza: Da li je voda univerzalni uslov za život?

U astrobiologiji se često traži tečna voda prilikom procene nastanjivosti drugih planeta i satelita. Razlog je jednostavan: svi oblici života koje poznajemo zavise od vode. Međutim, postavlja se pitanje da li je to zaista univerzalni uslov za život ili samo osobina zemaljske biohemije.

Jedna od mogućnosti koja se razmatra jeste postojanje života zasnovanog na amonijaku umesto vode. Amonijak je polaran molekul i može da rastvara određene organske supstance. Njegova prednost je što ostaje tečan na znatno nižim temperaturama od vode, pa bi mogao biti zanimljiv u hladnim sredinama.

Drugi primer je Titan, najveći Saturnov satelit. Na njegovoj površini postoje jezera i mora tečnih ugljovodonika, prvenstveno metana i etana. Iako su temperature na Titanu veoma niske, razmatra se mogućnost da bi u takvim uslovima mogla postojati potpuno drugačija biohemija od one koju poznajemo na Zemlji.

Razmatra se i koncentrovana sumporna kiselina kao potencijalni rastvarač, posebno u kontekstu Venerinih oblaka. Ona je veoma reaktivna i uništava mnoge organske molekule karakteristične za zemaljski život, ali istraživanja pokazuju da određena složena hemija u njoj ipak može biti moguća.

Ipak, ove hipoteze imaju ozbiljna ograničenja. Na primer, niske temperature na Titanu značajno usporavaju hemijske reakcije, a metan i etan imaju veoma ograničenu sposobnost rastvaranja mnogih supstanci. Amonijak takođe ima slabije vodonične veze od vode, dok bi život u sumpornoj kiselini zahtevao potpuno drugačiju biohemiju. Za sada nemamo dokaze da život zaista može funkcionisati u bilo kom od ovih alternativnih rastvarača.

Zbog toga je voda i dalje najperspektivniji rastvarač u potrazi za životom, ali ne možemo sa sigurnošću tvrditi da je jedini mogući.

```python
import requests
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from rdkit import Chem, rdBase
from rdkit.Chem import Descriptors, rdMolDescriptors, Draw
```

#### Preuzimanje CID i SMILES podataka iz PubChem-a:

```python
molekuli = [
    ("Glicin", "Aminokiselina", "glycine"),
    ("L-alanin", "Aminokiselina", "L-alanine"),
    ("L-serin", "Aminokiselina", "L-serine"),

    ("Adenin", "Nukleobaza", "adenine"),
    ("Guanin", "Nukleobaza", "guanine"),
    ("Uracil", "Nukleobaza", "uracil"),

    ("D-riboza", "Šećer", "D-ribose"),
    ("D-glukoza", "Šećer", "D-glucose"),

    ("AMP", "Nukleotid", "adenosine 5'-monophosphate"),
    ("ATP", "Nukleotid", "adenosine triphosphate")
]

def pubchem_podaci(naziv):

    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/name/{requests.utils.quote(naziv)}/cids/JSON"

    odgovor = requests.get(url, timeout=30)
    odgovor.raise_for_status()

    cid = odgovor.json()["IdentifierList"]["CID"][0]

    url = f"https://pubchem.ncbi.nlm.nih.gov/rest/pug/compound/cid/{cid}/property/SMILES/JSON"

    odgovor = requests.get(url, timeout=30)
    odgovor.raise_for_status()

    smiles = odgovor.json()["PropertyTable"]["Properties"][0]["SMILES"]

    return cid, smiles


podaci = []

for ime, klasa, pubchem_ime in molekuli:

    cid, smiles = pubchem_podaci(pubchem_ime)

    podaci.append({
        "Molekul": ime,
        "Klasa": klasa,
        "CID": cid,
        "SMILES": smiles
    })

df = pd.DataFrame(podaci)

display(df)
```

#### Izračunavanje molekulskih deskriptora pomoću RDKit-a:

```python
rezultati = []

molekulske_strukture = []

for _, red in df.iterrows():

    mol = Chem.MolFromSmiles(red["SMILES"])

    if mol is None:
        print("Greška pri učitavanju:", red["Molekul"])
        continue

    # Hiralni centri
    centri = Chem.FindMolChiralCenters(
        mol,
        includeUnassigned=True
    )

    broj_hiralnih = len(centri)

    definisani = sum(
        1 for atom, oznaka in centri
        if oznaka != "?"
    )

    # Deskriptori
    rezultati.append({
        "Molekul": red["Molekul"],
        "Klasa": red["Klasa"],
        "CID": red["CID"],
        "SMILES": red["SMILES"],
        "MW (g/mol)": Descriptors.MolWt(mol),
        "LogP": Descriptors.MolLogP(mol),
        "TPSA (Å²)": Descriptors.TPSA(mol),
        "HBD": Descriptors.NumHDonors(mol),
        "HBA": Descriptors.NumHAcceptors(mol),
        "Rotirajuće veze": Descriptors.NumRotatableBonds(mol),
        "Hiralni centri": broj_hiralnih,
        "Definisani hiralni centri": definisani
    })

    molekulske_strukture.append(mol)


df_rezultati = pd.DataFrame(rezultati)

display(df_rezultati.round(3))
```

#### Grafikon: molekulska masa u zavisnosti od TPSA:

```python
plt.figure(figsize=(10, 7))

for klasa in df_rezultati["Klasa"].unique():

    podskup = df_rezultati[
        df_rezultati["Klasa"] == klasa
    ]

    plt.scatter(
        podskup["MW (g/mol)"],
        podskup["TPSA (Å²)"],
        s=100,
        label=klasa
    )

    for _, red in podskup.iterrows():

        plt.annotate(
            red["Molekul"],
            (red["MW (g/mol)"], red["TPSA (Å²)"]),
            xytext=(5, 5),
            textcoords="offset points",
            fontsize=9
        )

plt.xlabel("Molekulska masa (g/mol)")
plt.ylabel("TPSA (Å²)")
plt.title("Molekulska masa u zavisnosti od polarnosti")

plt.legend(title="Klasa molekula")
plt.grid(alpha=0.3)

plt.show()
```

### Analiza molekulskih deskriptora
Izračunati molekulski deskriptori omogućavaju poređenje različitih klasa organskih molekula koji imaju značajnu ulogu u zemaljskom životu.

Jedna od očekivanih razlika odnosi se na molekulsku masu. Nukleotidi, poput AMP-a i ATP-a, imaju veću molekulsku masu od aminokiselina i nukleobaza, prvenstveno zbog prisustva šećera i fosfatnih grupa. Takođe, očekuje se da imaju veće vrednosti TPSA, što ukazuje na izraženiju polarnost. Šećeri imaju veliki broj hidroksilnih grupa, zbog čega mogu formirati veliki broj vodoničnih veza sa molekulima vode. Aminokiseline imaju amino i karboksilnu grupu, koje im omogućavaju različita hemijska svojstva.

LogP može pomoći u poređenju relativne sklonosti molekula da se rastvaraju u lipidnoj i vodenoj sredini, dok broj donora i akceptora vodoničnih veza daje dodatne informacije o njihovim mogućim interakcijama.

Ipak, ovi deskriptori sami po sebi ne mogu odrediti biološku funkciju ili poreklo molekula. Na primer, prisustvo adenina ne dokazuje da je molekul nastao biološkim putem, jer nukleobaze mogu nastati i abiotičkim hemijskim reakcijama.

Važno ograničenje predstavlja i stereohemija. Dva molekula mogu imati istu molekulsku formulu i veoma slične deskriptore, a različitu prostornu konfiguraciju. Takođe, protonacija molekula može promeniti njegov formalni naboj i neke izračunate osobine. Posebno je važno imati u vidu da aminokiseline i nukleotidi u vodenoj sredini mogu postojati u različitim protonacionim oblicima.

Konačno, uzorak od deset molekula relativno je mali, pa se dobijene razlike ne mogu generalizovati na sve aminokiseline, šećere, nukleobaze i nukleotide. Deskriptori su korisni za poređenje hemijskih struktura, ali nisu dovoljni za zaključivanje o biološkoj funkciji, nastanku ili prisustvu života.

### Zakljucak

Detekcija kompleksnih organskih molekula i izraženog enantiomernog viška može predstavljati potencijalnu biosignaturu, ali sama po sebi nije dovoljan dokaz postojanja života. Posebno je značajna hiralnost molekula: zemaljski organizmi uglavnom koriste L-aminokiseline za izgradnju proteina, pa bi izražena zastupljenost jednog enantiomera mogla ukazivati na biološku aktivnost.

Ipak, enantiomerni višak može nastati i abiotičkim procesima. Jedan od mogućih mehanizama je delovanje cirkularno polarizovanog ultraljubičastog zračenja, koje može selektivno razgrađivati jedan enantiomer. Takođe, kristalizacija i interakcije sa mineralnim površinama mogu doprineti stvaranju ili pojačavanju molekulske asimetrije.

Za pouzdaniju procenu bilo bi potrebno izmeriti izotopski sastav ugljenika, vodonika i azota, ispitati raspodelu enantiomera kod većeg broja aminokiselina i potražiti složenije molekule, poput nukleinskih kiselina ili njihovih polimera. Važno je isključiti i kontaminaciju uzoraka. Tek kombinacija više nezavisnih pokazatelja mogla bi pružiti ubedljiviji dokaz biološkog porekla. Prema tome, takav nalaz predstavlja razlog za dalje istraživanje, ali ne i potvrdu postojanja života.
