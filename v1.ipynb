#### 1. Pristrasnost metode detekcije

Tranzitna metoda lakše otkriva velike planete na bliskim orbitama, jer one češće prolaze ispred svoje zvezde i stvaraju veći pad sjaja. Male, stenovite planete na udaljenijim orbitama teže se otkrivaju.

Uticaj na listu: Kandidati koje možemo da detektujemo nisu reprezentativan uzorak svih planeta. Rang-lista može favorizovati velike planete i sisteme sa kratkim orbitalnim periodima, umesto planeta sličnih Zemlji. 

#### 2. Pristrasnost usled nepotpunih i nepreciznih podataka

Za neke planete nisu poznati svi parametri, poput mase, poluprečnika ili pouzdanih karakteristika matične zvezde. Ako analiza zahteva da svi parametri budu dostupni, određeni objekti biće isključeni. Ako se nedostajuće vrednosti procenjuju, rezultati mogu zavisiti od načina procene.

Uticaj na listu: Planete sa bolje izmerenim parametrima mogu dobiti prednost, dok se potencijalno zanimljivi kandidati sa velikim mernim neizvesnostima mogu izgubiti iz analize.

#### 3. Pristrasnost usled izbora kriterijuma i intervala ozračenja

Rangiranje zavisi od toga kako definišemo odgovarajući nivo zvezdanog ozračenja, na primer, granicama konzervativne ili optimističke nastanjive zone. Dodatni kriterijumi, poput dozvoljenog poluprečnika planete, mogu dalje promeniti uzorak.

Uticaj na listu: Planeta može biti odabrana ili odbačena samo zato što se nalazi blizu usvojene granice. Zato lista odražava i naše pretpostavke, a ne samo fizičke karakteristike planeta. Granice nastanjive zone zavise i od modela atmosfere i spektra matične zvezde.

### B. Izaberite alternativni interval ozračenja; ponovite izbor i kvantifikujte promenu broja kandidata.
Početni kriterijum bio je $(0.25\leq S\leq2.0,S_\oplus)$, uz poluprečnik do $(2R_\oplus)$ i udaljenost do 100 pc. Ovim postupkom izdvojeno je 45 kandidata iz 6451 zapisa.

U alternativnom scenariju interval je sužen na $(0.5\leq S\leq1.5,S_\oplus)$, dok su ostali kriterijumi zadržani. Dobijeno je 25 kandidata: 25 je zadržano, 20 odbačeno, a nijedan novi kandidat nije dodat.

Promena broja kandidata iznosi $\Delta N=25-45=-20,$

odnosno $\frac{-20}{45}\times100%$ = 44.4%

Dakle, sužavanje intervala ozračenja smanjilo je broj kandidata za 44,4%. Ovo pokazuje da je selekcija osetljiva na usvojene granice. Ipak, ozračenje samo po sebi ne određuje nastanjivost: atmosferski sastav, albedo i efekat staklene bašte mogu znatno promeniti površinske uslove.

### C. Fizicke degeneracije

#### 1. Degeneracija ozračenja i atmosferskih svojstava. 
Dve planete koje primaju jednako zračenje mogu imati različite površinske temperature zbog različitog albeda, atmosferskog pritiska i koncentracije gasova staklene bašte. Zato se samo na osnovu ozračenja ne može zaključiti da li na površini postoji tečna voda.

#### 2. Degeneracija mase, poluprečnika i sastava. 
Planete sličnih poluprečnika mogu imati različite mase i gustine, pa samim tim i različite unutrašnje strukture i atmosfere. Pošto je površinska gravitacija ($g=GM_p/R_p^2$), nepoznavanje mase ograničava procenu sposobnosti planete da zadrži atmosferu. Rang-lista zasnovana na poluprečniku, udaljenosti i ozračenju ne obuhvata ove razlike.

#### Predmet: Prioriteti za dodatna opažanja potencijalno nastanjivih egzoplaneta

#### Cilj i izbor kandidata. 
Predlažem da se vreme za dodatna opažanja prvenstveno usmeri na pet planeta: Proxima Centauri d, Proxima Centauri b, Ross 128 b, Teegarden's Star c i Teegarden's Star b. Ovaj izbor prati operativnu rang-listu dobijenu kombinovanjem udaljenosti i poluprečnika. Kandidati su relativno blizu Sunčevog sistema, a njihove vrednosti poluprečnika i ozračenja zadovoljavaju početne kriterijume analize.

#### Kriterijumi za raspodelu vremena. 
Prioritet treba određivati na osnovu četiri faktora: blizine sistema, verovatno stenovite prirode planete, pouzdanosti poznatih parametara i mogućnosti da dodatna opažanja razreše važne fizičke neizvesnosti. Blizina je važna jer može olakšati preciznija merenja i karakterizaciju atmosfere. Ipak, blizina i mali poluprečnik ne predstavljaju dokaz nastanjivosti. Pre konačne raspodele vremena potrebno je proveriti i sjaj matične zvezde, njenu aktivnost i opažljivost kandidata u odgovarajućem delu godine.

#### Ključne neizvesnosti. 
Trenutna rang-lista ne uključuje pouzdanu procenu sastava atmosfere, površinskog pritiska, albeda ni geološke aktivnosti. Posebno kod planeta oko crvenih patuljaka treba razmotriti zvezdane baklje, UV i rendgensko zračenje i moguće posledice po dugoročnu stabilnost atmosfere. Takođe, nesigurnost mase i poluprečnika može ograničiti procenu gustine i unutrašnje strukture.

#### Predložena naredna opažanja. 
Prvo bih zatražila dodatna precizna merenja mase metodom radijalnih brzina, gde su ona izvodljiva, kako bi se procenila gustina i bolje ograničio sastav planeta. Paralelno bih tražila dugoročnije praćenje aktivnosti matičnih zvezda. Za najpovoljnije sisteme sledeći korak bila bi procena mogućnosti tranzitne spektroskopije ili drugih tehnika karakterizacije atmosfere, u zavisnosti od geometrije sistema i očekivanog signala.

#### Zaključak. 
Predložena lista predstavlja početnu raspodelu prioriteta, a ne konačnu procenu nastanjivosti. Cilj dodatnih opažanja jeste smanjenje neizvesnosti i utvrđivanje koji kandidati zaista zaslužuju detaljnija istraživanja.
