## 1. Nastanjiva zona

Nastanjiva zona (engl. *Habitable Zone*, HZ) predstavlja oblast oko zvijezde u kojoj bi stjenovita planeta, pod odgovarajućim atmosferskim uslovima, mogla da održava vodu u tečnom stanju na svojoj površini.

Prisustvo tečne vode smatra se jednim od važnih preduslova za život kakav poznajemo.

Međutim, mogućnost postojanja tečne vode ne zavisi samo od udaljenosti planete od matične zvijezde.

Na klimatske uslove utiču:

- Luminoznost zvijezde
- Efektivna temperatura zvijezde
- Orbitalna udaljenost planete
- Ekscentricitet orbite
- Sastav i pritisak atmosfere
- Planetarni albedo
- Efekat staklene bašte
- Masa planete i evolucija njene atmosfere

### Unutrašnja granica nastanjive zone

Unutrašnja granica određena je uslovima pri kojima zvjezdano zračenje postaje toliko intenzivno da planeta može preći u stanje nekontrolisanog efekta staklene bašte (*runaway greenhouse*).

U tom slučaju površinski okeani mogu ispariti, što dodatno povećava količinu vodene pare u atmosferi i pojačava zagrijevanje.

### Spoljašnja granica nastanjive zone

Spoljašnja granica određena je uslovima pri kojima zagrijevanje atmosfere, čak i uz povećanu koncentraciju ugljen-dioksida, više nije dovoljno da održava tečnu vodu na površini planete.

### Važno

Ravnotežna temperatura ne uključuje u potpunosti uticaj atmosfere i efekta staklene bašte.

## 2. Luminoznost zvezde

Luminoznost predstavlja ukupnu energiju koju zvezda emituje u jedinici vremena.

Za zvezdu koju aproksimiramo crnim telom, luminoznost se može izračunati pomoću Stefan–Bolcmanovog zakona:

$$
L_\star = 4\pi R_\star^2 \sigma T_{\mathrm{eff}}^4
$$

gde su:

- $L_\star$ — luminoznost zvezde [W]
- $R_\star$ — poluprečnik zvezde [m]
- $\sigma$ — Stefan–Bolcmanova konstanta
- $T_{\mathrm{eff}}$ — efektivna temperatura zvezde [K]

Stefan–Bolcmanova konstanta iznosi:

$$
\sigma = 5.670374419 \times 10^{-8}
\ \mathrm{W\,m^{-2}\,K^{-4}}
$$

### Luminoznost u Sunčevim jedinicama

U astronomiji je često praktičnije koristiti odnos luminoznosti zvezde i Sunca:

$$
\boxed{
\frac{L_\star}{L_\odot}
=
\left(\frac{R_\star}{R_\odot}\right)^2
\left(\frac{T_{\mathrm{eff}}}{T_\odot}\right)^4
}
$$

Usvajamo:

$$
T_\odot = 5772\ \mathrm{K}
$$

Ovaj izraz nam omogućava da izračunamo luminoznost bez pretvaranja poluprečnika u metre.

### Zavisnost od temperature

Luminoznost zavisi od četvrtog stepena efektivne temperature:

$$
L_\star \propto T_{\mathrm{eff}}^4
$$

Zbog toga relativno mala promena temperature može značajno promeniti procenjenu luminoznost.

Za male promene:

$$
\frac{\Delta L}{L}
\approx
2\frac{\Delta R}{R}
+
4\frac{\Delta T}{T}
$$

Ovaj izraz opisuje približnu promenu luminoznosti pri malim promenama poluprečnika i temperature.

**Pitanje:**

## 3. Fluks zračenja koji planeta prima

Energija koju zvezda emituje raspoređuje se na sve veću površinu kako se udaljavamo od nje.

Ako pretpostavimo izotropno zračenje, fluks na udaljenosti $a$ iznosi:

$$
F = \frac{L_\star}{4\pi a^2}
$$

gde su:

- $F$ — fluks zračenja [W/m²]
- $L_\star$ — luminoznost zvezde [W]
- $a$ — udaljenost planete od zvezde [m]

Ovaj izraz predstavlja zakon opadanja fluksa sa kvadratom udaljenosti.

### Fluks u Zemljinim jedinicama

Definišimo relativni fluks:

$$
S = \frac{F}{F_\oplus}
$$

gde je $F_\oplus$ fluks koji Zemlja prima od Sunca na udaljenosti od 1 AU.

Dobijamo:

$$
\boxed{
S =
\frac{L_\star/L_\odot}
{(a/\mathrm{AU})^2}
}
$$

Za Zemlju:

$$
S_\oplus = 1
$$

Ako planeta prima dvostruko više zračenja od Zemlje:

$$
S = 2
$$

Ako prima upola manje:

$$
S = 0.5
$$

### Važna pretpostavka

Za kružnu orbitu udaljenost od zvezde je konstantna.

Za ekscentrične orbite fluks se menja tokom orbitalnog perioda, pa je potrebno razlikovati trenutni i vremenski usrednjeni fluks.

U ovoj vežbi prvo ćemo razmatrati približno kružne orbite.

**Pitanje:**

Ako planetu pomerimo sa 1 AU na 2 AU od zvezde, koliko puta će se smanjiti fluks koji prima?

## 4. Ravnotežna temperatura planete

Ravnotežna temperatura predstavlja temperaturu koju bi planeta imala kada bi energija koju apsorbuje bila jednaka energiji koju emituje u okolni prostor.

U jednostavnom modelu pretpostavljamo:

1. Planeta prima zračenje od svoje zvezde.
2. Deo zračenja se reflektuje.
3. Preostala energija se apsorbuje.
4. Planeta emituje energiju kao crno telo.
5. Toplota je ravnomerno raspoređena po celoj površini planete.
6. Atmosferski efekat staklene bašte nije eksplicitno uključen.

### Albedo

Bondov albedo, $A$, predstavlja udeo ukupnog upadnog elektromagnetnog zračenja koji planeta reflektuje.

$$
0 \leq A \leq 1
$$

Za $A=0$, planeta apsorbuje celokupno upadno zračenje.

Za $A=1$, planeta reflektuje celokupno upadno zračenje.

### Energetski bilans

Apsorbovana snaga je:

$$
P_{\mathrm{abs}} = \pi R_p^2 F(1-A)
$$

Emitovana snaga je:

$$
P_{\mathrm{emit}} = 4\pi R_p^2 \sigma T_{\mathrm{eq}}^4
$$

Iz uslova ravnoteže:

$$
P_{\mathrm{abs}} = P_{\mathrm{emit}}
$$

dobijamo:

$$
\boxed{
T_{\mathrm{eq}} =
\left(
\frac{F(1-A)}{4\sigma}
\right)^{1/4}
}
$$

### Ravnotežna temperatura i efekt staklene bašte

Ravnotežna temperatura nije isto što i površinska temperatura planete.

Na primer, za Zemlju:

$$
T_{\mathrm{eq},\oplus} \approx 255\ \mathrm{K}
$$

dok je prosečna površinska temperatura približno:

$$
T_{\mathrm{surface},\oplus} \approx 288\ \mathrm{K}
$$

Razlika je prvenstveno posledica atmosferskog efekta staklene bašte.

Zato se ravnotežna temperatura ne može direktno koristiti kao pouzdan kriterijum nastanjivosti.

**Pitanja:**

1. Zašto se u izrazu za apsorbovanu energiju pojavljuje $\pi R_p^2$, a za emitovanu energiju $4\pi R_p^2$?
2. Kako povećanje albeda utiče na ravnotežnu temperaturu?
3. Zašto $T_{\mathrm{eq}} < 273$ K ne znači da tečna voda ne može postojati na površini planete?

## 5. Model nastanjive zone — Kopparapu et al. (2014)

Granice nastanjive zone zavise od luminoznosti zvezde, ali i od
njenog spektralnog energetskog rasporeda, odnosno efektivne
temperature.

Planete koje kruže oko zvezda različitih temperatura ne
apsorbuju zvezdano zračenje na isti način, čak i ako primaju
jednaku ukupnu količinu energije.

Na primer, hladnije zvezde emituju veći deo energije na dužim
talasnim dužinama, što utiče na apsorpciju zračenja u atmosferi
i na površini planete.

Kopparapu et al. (2014) razvili su model za određivanje
granica nastanjive zone oko zvezda glavnog niza.

### Efektivni zvezdani fluks

Granice nastanjive zone izražavaju se preko efektivnog
zvezdanog fluksa:

$$
\boxed{
S_{\mathrm{eff}} =
S_{\mathrm{eff},\odot}
+ aT' + bT'^2 + cT'^3 + dT'^4
}
$$

gde je:

$$
T' = T_{\mathrm{eff}} - 5780\ \mathrm{K}
$$

Koeficijenti:

$$
S_{\mathrm{eff},\odot},\ a,\ b,\ c,\ d
$$

zavise od izabrane granice nastanjive zone.

### Oblast primene modela

Model se primenjuje za efektivne temperature zvezda:

$$
\boxed{
2600\ \mathrm{K} \leq T_{\mathrm{eff}} \leq 7200\ \mathrm{K}
}
$$

Van ovog intervala polinom ne treba koristiti bez dodatnog
opravdanja, jer bi to predstavljalo ekstrapolaciju modela.

### Granice nastanjive zone

Kada znamo efektivni fluks određene granice i luminoznost
zvezde, odgovarajuća orbitalna udaljenost iznosi:

$$
\boxed{
d_{\mathrm{HZ}} =
\sqrt{
\frac{L_\star/L_\odot}
{S_{\mathrm{eff}}}
}\ \mathrm{AU}
}
$$

Veći granični fluks odgovara manjoj orbitalnoj udaljenosti,
odnosno unutrašnjoj granici nastanjive zone.

Manji granični fluks odgovara većoj udaljenosti, odnosno
spoljašnjoj granici.

### Pitanja

3. Kako bi se promenile granice nastanjive zone kada bi se
   luminoznost zvezde povećala četiri puta, a efektivna
   temperatura ostala ista?

## 6. Različite definicije nastanjive zone

Granice nastanjive zone nisu univerzalne.

U literaturi se često razlikuju konzervativna i optimistična
nastanjiva zona.

### Konzervativna nastanjiva zona

Konzervativna nastanjiva zona definisana je pomoću:

**1. Runaway Greenhouse — unutrašnja granica**

Granica iznad koje se u pretpostavljenom klimatskom modelu
može javiti nekontrolisani efekat staklene bašte.

**2. Maximum Greenhouse — spoljašnja granica**

Granica pri kojoj dodatno povećanje koncentracije CO2 više
ne omogućava efikasnije zagrevanje planete, jer efekti
rasejanja i refleksije postaju sve važniji.

### Optimistična nastanjiva zona

Optimistična nastanjiva zona koristi šire, empirijski
motivisane granice:

**1. Recent Venus — unutrašnja granica**

Granica zasnovana na pretpostavkama o tome kada je Venera
mogla izgubiti uslove pogodne za postojanje površinske vode.

**2. Early Mars — spoljašnja granica**

Granica zasnovana na geološkim dokazima da je na ranom
Marsu postojala tečna voda.

### Poređenje

Konzervativna zona:

$$
d_{\mathrm{RG}} \leq a \leq d_{\mathrm{MG}}
$$

Optimistična zona:

$$
d_{\mathrm{RV}} \leq a \leq d_{\mathrm{EM}}
$$

gde su:

- RG — Runaway Greenhouse
- MG — Maximum Greenhouse
- RV — Recent Venus
- EM — Early Mars

Optimistična zona je šira od konzervativne.

### Važno

Izbor granica predstavlja izbor klimatskog modela i njegovih
pretpostavki, a ne dodatnu mernu neizvesnost.

U ovoj vežbi koristićemo koeficijente za planetu mase
1 Zemljine mase iz rada Kopparapu et al. (2014).

Masa planete može uticati na granice nastanjive zone,
ali taj efekat nećemo detaljnije modelovati.

## 8. Merne greske

Do sada smo pretpostavljali da su poluprečnik zvezde,
efektivna temperatura i orbitalna udaljenost poznati
sa potpunom preciznošću.

U stvarnosti, sve ove veličine određuju se na osnovu
posmatranja i imaju određene merne neizvesnosti.

Na primer:

$$
R_\star=(0.90\pm0.04)\,R_\odot
$$

$$
T_{\mathrm{eff}}=(5400\pm100)\,\mathrm{K}
$$

$$
a=(0.90\pm0.03)\,\mathrm{AU}
$$

Centralne vrednosti predstavljaju procene parametara,
dok neizvesnosti opisuju ograničenu preciznost merenja.

### Propagacija greske

Luminoznost zvezde zavisi od poluprečnika i temperature:

$$
L_\star \propto R_\star^2 T_{\mathrm{eff}}^4
$$

Ako su greske male, približno simetrične i međusobno
nezavisne, relativna standardna greska luminoznosti je:

$$
\boxed{
\left(\frac{\sigma_L}{L}\right)^2
\approx
\left(2\frac{\sigma_R}{R}\right)^2
+
\left(4\frac{\sigma_T}{T}\right)^2
}
$$

Slično, pošto je:

$$
S\propto La^{-2}
$$

dobijamo:

$$
\boxed{
\left(\frac{\sigma_S}{S}\right)^2
\approx
\left(\frac{\sigma_L}{L}\right)^2
+
\left(2\frac{\sigma_a}{a}\right)^2
}
$$

Ovi izrazi predstavljaju linearizovanu propagaciju
greske.

Oni nisu potpuno pouzdani kada su greske velike,
raspodele izrazito asimetrične ili parametri međusobno
korelisani.

### Nedostajuće greska

Ako za neki parametar nije navedena greska,
to ne znači da je njegova vrednost potpuno tačna.

Nedostajuće greske ne smemo automatski
zameniti nulom.

Umesto toga potrebno je:

- Evidentirati nedostajuće podatke.
- Navesti dodatne pretpostavke.
- Ispitati koliko rezultat zavisi od tih pretpostavki.

## 9. Raspodele verovatnoće

Umesto da merenu veličinu predstavljamo jednom vrednošću,
možemo je opisati raspodelom verovatnoće.

### Normalna raspodela

Jedna od najčešće korišćenih raspodela je normalna
(Gausova) raspodela:

$$
p(x)=
\frac{1}{\sigma\sqrt{2\pi}}
\exp\left[
-\frac{(x-\mu)^2}{2\sigma^2}
\right]
$$

gde su:

- $\mu$ — srednja vrednost
- $\sigma$ — standardna devijacija

Na primer:

$$
T_{\mathrm{eff}}\sim\mathcal{N}(5400,100^2)
$$

To znači da efektivnu temperaturu modelujemo normalnom
raspodelom sa srednjom vrednošću 5400 K i standardnom
devijacijom 100 K.

### Fizička ograničenja

Normalna raspodela matematički dozvoljava i negativne
vrednosti, što nije fizički prihvatljivo za mnoge veličine.

Na primer:

$$
R_\star>0,\qquad T_{\mathrm{eff}}>0,\qquad a>0
$$

Zato se u numeričkim simulacijama mogu koristiti:

- Odsečena normalna raspodela
- Lognormalna raspodela
- Druge raspodele sa fizički dozvoljenim opsegom

Izbor raspodele treba da bude zasnovan na dostupnim
podacima i pretpostavkama analize.



## 10. Monte Karlo metoda

Monte Karlo metoda predstavlja numerički pristup u kojem
generišemo veliki broj slučajnih realizacija parametara
da bismo ispitali ponašanje modela.

Umesto jednog skupa vrednosti:

$$
R_\star,\ T_{\mathrm{eff}},\ a
$$

generišemo veliki broj mogućih kombinacija:

$$
(R_1,T_1,a_1)
$$

$$
(R_2,T_2,a_2)
$$

$$
\vdots
$$

$$
(R_N,T_N,a_N)
$$

Za svaku realizaciju izračunavamo:

$$
L_i,\ S_i,\ T_{\mathrm{eq},i}
$$

kao i granice nastanjive zone:

$$
d_{\mathrm{inner},i},
\quad d_{\mathrm{outer},i}
$$

Na taj način, umesto jednog rezultata,
dobijamo raspodelu mogućih rezultata.

### Broj realizacija

U ovoj vežbi koristićemo:

$$
N=50\,000
$$

Monte Karlo realizacija.

Veći broj realizacija obično smanjuje numerički šum
procene, ali ne uklanja sistematske greške niti
pogrešne pretpostavke modela.

### Reproduktivnost

Da bismo prilikom svakog pokretanja programa mogli
da dobijemo iste pseudoslučajne brojeve, koristićemo
fiksirano početno stanje generatora:
    rng = np.random.default_rng(42)

Broj 42 predstavlja izabrani seed.

Fiksiranje seed-a omogućava reprodukovanje numeričke
simulacije, što je važno za naučna istraživanja.

### Prikaz rezultata

Monte Karlo rezultate možemo predstaviti pomoću:

- Histograma
- Medijane raspodele
- Percentila
- Intervalnih procena

Na primer, interval između 16. i 84. percentila
često predstavlja praktičan opis centralnog dela raspodele.

Ne treba ga automatski poistovećivati sa univerzalnim
intervalom pouzdanosti.

2. Zašto je važno postaviti seed generatora?
3. Da li povećanje broja simulacija može ukloniti
   pogrešne pretpostavke o klimatskom modelu?

## 11. Procena pripadnosti nastanjivoj zoni

Za svaku Monte Karlo realizaciju proveravamo da li se
planeta nalazi između unutrašnje i spoljašnje granice
nastanjive zone.

Uslov je:

$$
d_{\mathrm{inner},i}
\leq a_i
\leq d_{\mathrm{outer},i}
$$

Ekvivalentno, možemo porediti flukseve:

$$
S_{\mathrm{outer},i}
\leq S_i
\leq S_{\mathrm{inner},i}
$$

Definišimo indikator:

$$
I_i=
\begin{cases}
1, & \text{ako se planeta nalazi unutar HZ}\\
0, & \text{ako se planeta nalazi izvan HZ}
\end{cases}
$$

Udeo realizacija unutar nastanjive zone iznosi:

$$
\boxed{
\widehat{P}_{\mathrm{HZ}}=
\frac{1}{N}\sum_{i=1}^{N} I_i
}
$$

Ova veličina predstavlja procenjeni udeo konfiguracija
koje zadovoljavaju kriterijum izabranog modela.

Njeno tumačenje zavisi od:

- Izabranih raspodela ulaznih parametara
- Uključenih mernih neizvesnosti
- Pretpostavki o njihovoj međusobnoj zavisnosti
- Izabranih granica nastanjive zone
- Primenjenog klimatskog modela

### Važno upozorenje

$$
P_{\mathrm{HZ}} \neq P(\mathrm{život})
$$

Čak i ako dobijemo:

$$
P_{\mathrm{HZ}}=0.95
$$

to ne znači da na planeti postoji 95% verovatnoće
za postojanje života.

To samo znači da 95% simuliranih konfiguracija
ispunjava uslov položaja unutar izabrane
nastanjive zone.

### Pitanje

## 12. Uticaj klimatskih pretpostavki

Do sada smo razmatrali merne neizvesnosti.

Međutim, rezultat zavisi i od pretpostavki koje
koristimo za opis planetarne klime.

### Uticaj albeda

Ravnotežna temperatura zavisi od albeda:

$$
T_{\mathrm{eq}}\propto(1-A)^{1/4}
$$

Veći albedo znači da planeta reflektuje veći deo
upadnog zračenja i ima nižu ravnotežnu temperaturu
u jednostavnom energetskom modelu.

U vežbi ćemo uporediti tri primera:

$$
A=0.1,\quad A=0.3,\quad A=0.6
$$

### Uticaj atmosfere

Stvarna površinska temperatura zavisi od atmosferskog
sastava, pritiska, oblaka, transporta toplote i drugih
procesa.

Ravnotežna temperatura sama po sebi ne opisuje
te procese.

Za ilustraciju možemo koristiti pojednostavljeni
model sa parametrizovanim dodatnim zagrevanjem:

$$
T_{\mathrm{surface}}=
T_{\mathrm{eq}}+\Delta T_{\mathrm{GH}}
$$

gde je $\Delta T_{\mathrm{GH}}$ pretpostavljeni doprinos
efekta staklene bašte.

Ovaj izraz nije potpuni fizički klimatski model
i služi isključivo kao ilustracija osetljivosti
rezultata na atmosferske pretpostavke.

### Važna razlika

Promena albeda ili pretpostavljenog efekta staklene
bašte menja procenu ravnotežne ili površinske
temperature.

Međutim, time ne menjamo automatski granice
Kopparapu modela, jer one već proizlaze iz
određenih klimatskih pretpostavki.

Za drugačiju atmosferu potreban je odgovarajući
klimatski model.
