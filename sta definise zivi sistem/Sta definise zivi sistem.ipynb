# ŠTA DEFINIŠE ŽIVI SISTEM?


### 1. Šta je život i zašto ga je teško definisati?

Život je teško jednoznačno definisati jer postoje sistemi koji poseduju neke osobine živih bića, ali se ipak ne smatraju živim. Metabolička definicija posmatra život kao sistem koji koristi i transformiše energiju radi održavanja i rasta. Međutim, i plamen troši energiju i širi se, a nije živ. Termodinamička definicija naglašava održavanje lokalnog reda uz razmenu energije sa okolinom, ali bi prema njoj i kristali mogli biti smatrani živim. Biohemijska definicija zasniva se na prisustvu karakterističnih molekula, poput proteina i nukleinskih kiselina, ali možda isključuje oblike života zasnovane na drugačijoj hemiji. Darvinistička definicija život posmatra kao sistem sposoban za evoluciju prirodnom selekcijom, mada isključuje viruse koji ne mogu samostalno da se razmnožavaju. Za potragu za životom na drugim planetama najkorisniji je kombinovani pristup, jer nijedan pojedinačni kriterijum nije dovoljan za pouzdanu identifikaciju.

### 2. Šta su nam pokazali Viking eksperimenti na Marsu?

Misije Viking iz sedamdesetih godina prošlog veka predstavljale su prvi ozbiljan pokušaj direktnog traženja života na Marsu. Eksperiment obeleženog oslobađanja (Labeled Release) imao je zadatak da utvrdi da li mikroorganizmi u marsovskom zemljištu obavljaju metabolizam. Uzorku zemljišta dodavan je hranljivi rastvor sa radioaktivno obeleženim ugljenikom, a zatim se pratilo oslobađanje radioaktivnih gasova kao mogućeg znaka biološke aktivnosti. Rezultati su pokazali signal koji je mogao ukazivati na prisustvo života. Međutim, GCMS instrument nije detektovao organske molekule, što je značajno uticalo na zaključak da signal verovatno ima nebiološko poreklo. Ipak, odsustvo detekcije organskih materija nije definitivno dokazivalo odsustvo života. Interpretaciju bi poboljšalo ponavljanje eksperimenta na prethodno sterilisanom uzorku, kao i nezavisna analiza gasova i organskih jedinjenja.

### 3. Koje su granice poznatog života?

Za život kakav poznajemo na Zemlji neophodni su tečna voda, izvor energije, ugljenik i drugi hemijski elementi, poput azota, sumpora i fosfora. Voda omogućava odvijanje biohemijskih reakcija, dok energija pokreće metabolizam, a hemijski elementi učestvuju u izgradnji ćelijskih struktura. Jedan od ograničavajućih parametara je temperatura: poznati organizmi mogu aktivno da rastu u približnom rasponu od −15 °C do 122 °C. Ekstremno visoke temperature narušavaju stabilnost proteina i drugih biomolekula. Drugi važan parametar je aktivnost vode, koja opisuje koliko je vode dostupno za biohemijske procese. Pri veoma niskoj aktivnosti vode metabolizam postaje otežan ili prestaje. Važno je razlikovati preživljavanje i aktivan rast: ekstremofili mogu preživeti nepovoljne uslove, ali to ne znači da u njima mogu da se razmnožavaju.

### 4. Zašto je nastanjivost planetarno svojstvo?

Nastanjivost zavisi od međusobnog uticaja različitih osobina planete, a ne od samo jednog faktora. Kao prvi primer može se navesti atmosfera. Ona omogućava zadržavanje toplote i tečne vode, a ujedno štiti površinu od dela štetnog zračenja i manjih meteoroida. Međutim, previše gusta atmosfera ili izražen efekat staklene bašte mogu dovesti do ekstremno visokih temperatura. Drugi primer je magnetno polje, koje može smanjiti uticaj naelektrisanih čestica zvezdanog vetra i usporiti eroziju atmosfere. Ipak, njegovo odsustvo ne znači da život nije moguć, jer atmosfera sama može pružiti dovoljnu zaštitu, a život može postojati i ispod površine. Zbog toga nijedan od ovih faktora nije univerzalno neophodan: različite kombinacije planetarnih uslova mogu omogućiti postojanje života.

Izvor: Laskar, J., Joutel, F., & Robutel, P. (1993), Stabilization of the Earth's obliquity by the Moon.  
Ilustracija: Figure 4.8 iz lekcije 4. Life and its Requirements – ASTR3800 Astrobiology.
#### Opis ilustracije
Ilustracija prikazuje moguće promene nagiba Zemljine ose rotacije (obliquity) u zavisnosti od konstante precesije, uz poređenje Zemlje sa Mesecom i hipotetičke Zemlje bez Meseca. Na horizontalnoj osi prikazan je nagib ose rotacije u stepenima, dok leva vertikalna osa predstavlja konstantu precesije u lučnim sekundama godišnje. Desna vertikalna osa prikazuje vremensku skalu u milijardama godina (Ga). Boje predstavljaju različite veličine promena nagiba: plava oblast odgovara malim promenama, dok crvena ukazuje na mogućnost znatno većih varijacija. Za današnju Zemlju, sa Mesecom, nagib ose iznosi približno 23,3°, a konstanta precesije oko 51 lučnu sekundu godišnje. Nasuprot tome, model Zemlje bez Meseca pokazuje da bi nagib ose mogao da varira i do približno 60°.
#### Naučni značaj ilustracije
Ilustracija podržava tvrdnju da Mesec ima važnu ulogu u stabilizaciji nagiba Zemljine ose, a time posredno i u dugoročnoj stabilnosti klime. Velike promene nagiba ose mogle bi da izazovu značajne promene u raspodeli Sunčevog zračenja između različitih geografskih širina i godišnjih doba. To bi moglo dovesti do izraženijih klimatskih oscilacija i otežati održavanje stabilnih uslova za pojedine ekosisteme. Ipak, odsustvo velikog prirodnog satelita ne znači automatski da planeta ne može biti nastanjiva, jer stabilnost nagiba zavisi i od brzine rotacije i gravitacionih uticaja drugih planeta. Zato je Mesec važan faktor koji može doprineti nastanjivosti, ali nije univerzalni preduslov za postojanje života.
